```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.metrics import r2_score, mean_absolute_percentage_error
from xgboost import XGBRegressor
```

Carregamento de Dados

```python
df = pd.read_csv(
    r"data\dados_preprocessados_temp_rj.csv",
    index_col="time",
    parse_dates=True
)

df
```

Exploração Inicial

```python
display(df[["TARGET"]].describe())

df[["TARGET"]].plot(figsize=(12, 4))
```

```python
sns.histplot(df[["TARGET"]], kde=True)
```

Preparação de Dados para treinamento

```python
df_processado = df.copy()
```

```python
df_processado.head(5)
```

Separação de Variáveis TARGET (y) e Features (x)

```python
y_target = df_processado[["TARGET"]]

x_features = df_processado.drop(columns=["TARGET"])
```

Separação em Treino e Teste

```python
#Separação 70% para treino e 30% para teste

#Primeiros 25 anos (1991 - 2015)

y_train = y_target["1991":"2015"]
x_train = x_features["1991":"2015"]

#Últimos 10 anos (2016 - 2025)

y_test = y_target["2016":"2025"]
x_test = x_features["2016":"2025"]

#Vizualizando percentuais
pct_treino = (len(y_train))/(len(y_target)) * 100
print(f"Percentual Dados Treino: {pct_treino:.0f}%")

pct_teste = (len(y_test))/(len(y_target)) * 100
print(f"Percentual Dados Teste: {pct_teste:.0f}%")
```

Treinamento do Modelo

```python
#Modelo escolhido -> XGBoost -> Modelo Gradiente

modelo = XGBRegressor(
    random_state=42,
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    colsample_bytree=0.8,
    subsample=0.8,
    reg_alpha=0.1,
    reg_lambda=0.1,
)

#modelo_parameterizado

modelo.fit(x_train, y_train)
```

```python
#Buscar previsões nas amostras de treino

pred_train = pd.DataFrame(modelo.predict(x_train), index=x_train.index, columns=["Y_PRED"])
pred_test = pd.DataFrame(modelo.predict(x_test), index=x_test.index, columns=["Y_PRED"])

pred_test
```

```python
fig, ax = plt.subplots(figsize=(12,4))

ax.plot(y_target, label="Valores Reais")
ax.plot(pred_train, label="Previsões - Treino")
ax.plot(pred_test, label="Previsões - Teste")

ax.legend(ncols=3)
```

```python
#R²
r2_test = r2_score(y_true=y_test, y_pred=pred_test)
r2_train = r2_score(y_true=y_train, y_pred=pred_train)

print(f"-Valores de R² -> Treino: {r2_train:.2f} | Teste: {r2_test:.2f}")

#MAPE

mape_test = mean_absolute_percentage_error(y_true=y_test, y_pred=pred_test) * 100
mape_train = mean_absolute_percentage_error(y_true=y_train, y_pred=pred_train) * 100

print(f"-Valores de MAPE -> Treino: {mape_train:.2f} | Teste: {mape_test:.2f}")
```

```python
# Interpretação:
# - R² de 0,84 no teste: o modelo explica ~84% da variação da temperatura.
# - MAPE de ~3%: o erro médio é de cerca de 3% da temperatura real
#   (~0,7 °C numa média de 24 °C).
# - Há uma queda do treino para o teste (R² 0,96 -> 0,84; MAPE dobra).
```
